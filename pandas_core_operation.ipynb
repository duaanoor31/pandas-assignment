```python
#Student Practice Task:
#Task 1
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

average = df["Marks"].mean()

maximum = df["Marks"].max()

minimum = df["Marks"].min()

total_students = df["Name"].count()

print("Average Marks:", average)
print("Maximum Marks:", maximum)
print("Minimum Marks:", minimum)
print("Total Students:", total_students)
```

```python
#Task 2
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)
print("Students whose marks are greater than or equal to 50:")
print(df[df["Marks"] >= 50])

print("\nStudents whose marks are less than 50:")
print(df[df["Marks"] < 50])
```

```python
#Task 3
df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Zain'],
    'Marks': [70, 85, 60, 95, 45],
    'City': ['Karachi', 'Lahore', 'Islamabad', 'Multan', 'Peshawar']
})

print("All Students:")
print(df)

karachi_students = df[df['City'] == 'Karachi']

print("\nStudents from Karachi:")
print(karachi_students)
```

```python
#Task 4:
df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Zain'],
    'Course': ['AI', 'Web', 'AI', 'Web', 'AI'],
    'Marks': [70, 85, 60, 95, 80]
})

print("All Students:")
print(df)

average_marks = df.groupby('Course')['Marks'].mean()

print("\nAverage Marks of Each Course:")
print(average_marks)
```

```python
#Task 5:
df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Zain'],
    'Marks': [80, None, 70, None, 90]
})

print("Number of missing values:")
print(df['Marks'].isnull().sum())

average = df['Marks'].mean()

df['Marks'] = df['Marks'].fillna(average)

print("\nCleaned DataFrame:")
print(df)
```
